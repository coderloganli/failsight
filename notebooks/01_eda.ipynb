# Exploratory analysis of Backblaze SMART telemetry

Skeleton notebook. Run `failsight download` and `failsight build` first so that `data/failsight.duckdb` contains the `clean` and `features` tables.

Questions this notebook answers:

1. How are the SMART attributes distributed for drives that failed versus drives that did not?
2. How do failure rates differ across drive models (and which models have too few failures for supervised training)?
3. How do attributes evolve in the days before a failure, and which rolling-window deltas and rates of change are worth using as features?

```python
from failsight import eda
from failsight.config import load_config, smart_columns
from failsight.data import connect

cfg = load_config()
con = connect(cfg["paths"]["database"], read_only=True)
```

## 1. Attribute distributions: failed vs healthy drives

```python
sample = eda.sample_drives(con, n_drives=2000)
eda.plot_attribute_distributions(sample, smart_columns(cfg));
```

## 2. Failure rates by drive model

```python
rates = eda.failure_rates_by_model(con, min_drive_days=10_000)
eda.plot_failure_rates(rates, top_n=20);
```

## 3. Pre-failure trajectories

```python
for column in ["smart_5_raw", "smart_187_raw", "smart_197_raw"]:
    traj = eda.prefailure_trajectories(con, column, days_before=60)
    eda.plot_prefailure_trajectory(traj, column)
```
